```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

sum = num1 + num2
diff = num1 - num2
prod = num1 * num2
quotient = num1 / num2
floordiv = num1 // num2
modulus = num1 % num2

print("Sum:", sum)
print("Difference:", diff)
print("Product:", prod)
print("Quotient:", quotient)
print("Floor division:", floordiv)
print("Modulus:", modulus)
```

```python
temp_in_fahrenheit = input("Enter temperature in Fahrenheit: ")
temp_in_celsius = (int(temp_in_fahrenheit) - 32) * (5/9)
print(f"Temperature in fahrenheit {temp_in_fahrenheit} and in celsius {temp_in_celsius:.0f}")
```

```python
print("Prime numbers: ")
for num in range(1, 101):
    if num > 1:
        for i in range(2, num):
            if (num % i) == 0:
                break
        else:
            print(num, end=" ")   
```

```python
num = int(input("Enter a number: "))
if num & 1:
    print(num,"is odd")
else:
    print(num,"is even")
```
